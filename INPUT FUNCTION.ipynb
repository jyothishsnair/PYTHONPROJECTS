```python
while(True):
    try:
        r = float(input("Enter Radius:"))
        if(r==0):
            print("Invaild Radius")
        elif(r<0):
            print("Invaild Radius")
        else:
            ac = 3.14*r*r
            print("Area of Circle=",ac)
    except ValueError:
          print("Don't Enter ALNUMNS,STRS AND SYMBOLS- TRY AGAIN")
    
```
